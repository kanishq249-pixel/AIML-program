```python
import math
import numpy as np
import scipy
from scipy import stats
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python



print("MATH LIBRARY")
print("Square root of 25:", math.sqrt(25))
print("Factorial of 5:", math.factorial(5))
print("Power of 2:", math.pow(2, 3))


print("\nNUMPY LIBRARY")
numbers = np.array([10, 20, 30, 40, 50])
print("Array:", numbers)
print("Mean:", np.mean(numbers))
print("Maximum:", np.max(numbers))
print("Sum:", np.sum(numbers))


print("\nSCIPY LIBRARY")
data = [10, 20, 30, 40, 50]
print("Mode:", stats.mode(data))
print("SciPy Version:", scipy.__version__)


print("\nPANDAS LIBRARY")
df = pd.DataFrame({
    "Name": ["Amit", "Riya", "Neha", "Raj"],
    "Marks": [75, 85, 90, 80]
})
print(df)
print("Average Marks:", df["Marks"].mean())


plt.plot(df["Name"], df["Marks"], marker="o")
plt.title("Student Marks")
plt.xlabel("Student Name")
plt.ylabel("Marks")
plt.show()

sns.barplot(data=df, x="Name", y="Marks")
plt.title("Student Marks Using Seaborn")
plt.show()
```
