```python
graph = {
    'A': ['B','D'],
    'B': ['C','E'],
    'C': [],
    'D': ['F'],
    'E': [],
    'F': []
}

def dfs(graph,start):
    visited = set()
    stack = [start]

    while stack:
      node = stack.pop()

      if node not in visited:
       print(node,end="")

     # Add neighbor in reverse order
       for neighbor in reversed(graph[node]):
          if neighbor not in visited:
            stack.append(neighbor)

       visited.add(node)

dfs(graph,'A')

```
